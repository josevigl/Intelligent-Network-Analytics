<h3 style="color: #8AB9FF; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
01 · NETWORK ANOMALY DETECTION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
Telecommunications networks continuously generate large volumes of performance measurements that reflect their operational conditions. However, network degradation, unexpected traffic patterns, interference and equipment malfunctions can produce abnormal behaviours that compromise service quality.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The first objective is to identify abnormal network behaviour and distinguish between normal operating conditions and anomalous situations. Early anomaly detection is essential for improving network observability, reducing service disruptions and maintaining reliable connectivity.
</p>

<div style="border-top: 1px solid #333333; margin: 24px 0;"></div>

<h3 style="color: #FF8BAF; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
02 · ANOMALY TYPE CLASSIFICATION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
Not all network anomalies have the same origin or operational impact. Different situations, such as radio interference, network congestion, abnormal traffic behaviour or connectivity degradation, can affect telecommunications infrastructure in different ways.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The second objective is to distinguish between different categories of network anomalies. Understanding the nature of each abnormal event provides valuable information for network diagnosis, incident prioritisation and the identification of recurring operational problems.
</p>

<div style="border-top: 1px solid #333333; margin: 24px 0;"></div>

<h3 style="color: #8AB9FF; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
03 · TEMPORAL ANOMALY LOCALIZATION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
Network anomalies are dynamic events that evolve over time. Their duration, onset and recovery are important characteristics when analysing the behaviour of telecommunications infrastructure. Identifying the presence of an anomaly alone does not provide a complete understanding of the incident.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The third objective is to determine the precise temporal interval in which anomalous behaviour occurs. This enables a more detailed understanding of how network incidents develop, how long they persist and when normal operating conditions are restored.
</p>

```python
ds = load_dataset("AliMaatouk/TelecomTS")
df = ds["train"].to_pandas()
df
```

```python
df_true = df[df['anomalies'].apply(lambda x: x['exists'] == True)]
df_true
```

```python
unique_durations = df_true['anomalies'].apply(
    lambda x: (
        x.get('anomaly_duration', {}).get('start'),
        x.get('anomaly_duration', {}).get('end')
    )
).unique()

print(unique_durations)
```

```python
from collections import Counter

anomaly_types = df_true['anomalies'].apply(
    lambda x: x.get('anomaly_duration')
)

print(anomaly_types.value_counts(dropna=False))
```

```python
duration = df_true['anomalies'].apply(
    lambda x: x.get('anomaly_duration')
)

full_window = duration.apply(
    lambda x: x.get('start') == 0 and x.get('end') == 127
)

print("Ventanas completas:", full_window.sum())
print("Ventanas parciales:", (~full_window).sum())

print("Porcentaje completas:", full_window.mean() * 100)
print("Porcentaje parciales:", (~full_window).mean() * 100)
```

```python
durations = duration.apply(
    lambda x: x.get('end') - x.get('start') + 1
)

print(durations.describe())
```

```python
starts = duration.apply(lambda x: x.get('start'))
ends = duration.apply(lambda x: x.get('end'))

print("Distribución de start:")
print(starts.value_counts().sort_index())

print("\nDistribución de end:")
print(ends.value_counts().sort_index())
```

```python
df_analysis = df_true.copy()

df_analysis['anomaly_type'] = df_analysis['anomalies'].apply(
    lambda x: x.get('type')
)

df_analysis['start'] = df_analysis['anomalies'].apply(
    lambda x: x.get('anomaly_duration').get('start')
)

df_analysis['end'] = df_analysis['anomalies'].apply(
    lambda x: x.get('anomaly_duration').get('end')
)

df_analysis['duration'] = (
    df_analysis['end'] - df_analysis['start'] + 1
)

df_analysis.groupby('anomaly_type')['duration'].agg(
    ['count', 'mean', 'median', 'min', 'max']
).sort_values('mean', ascending=False)
```

```python
df_analysis['is_full_window'] = (
    (df_analysis['start'] == 0) &
    (df_analysis['end'] == 127)
)

pd.crosstab(
    df_analysis['anomaly_type'],
    df_analysis['is_full_window'],
    margins=True
)
```
