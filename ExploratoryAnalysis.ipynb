<h3 style="color: #004D98; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
01 · NETWORK ANOMALY DETECTION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
Telecommunications networks continuously generate large volumes of performance measurements that reflect their operational conditions. However, network degradation, unexpected traffic patterns, interference and equipment malfunctions can produce abnormal behaviours that compromise service quality.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The first objective is to identify abnormal network behaviour and distinguish between normal operating conditions and anomalous situations. Early anomaly detection is essential for improving network observability, reducing service disruptions and maintaining reliable connectivity.
</p>

<div style="border-top: 1px solid #333333; margin: 24px 0;"></div>

<h3 style="color: #A50044; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
02 · ANOMALY TYPE CLASSIFICATION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
Not all network anomalies have the same origin or operational impact. Different situations, such as radio interference, network congestion, abnormal traffic behaviour or connectivity degradation, can affect telecommunications infrastructure in different ways.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The second objective is to distinguish between different categories of network anomalies. Understanding the nature of each abnormal event provides valuable information for network diagnosis, incident prioritisation and the identification of recurring operational problems.
</p>

<div style="border-top: 1px solid #333333; margin: 24px 0;"></div>

<h3 style="color: #004D98; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
03 · TEMPORAL ANOMALY LOCALIZATION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
Network anomalies are dynamic events that evolve over time. Their duration, onset and recovery are important characteristics when analysing the behaviour of telecommunications infrastructure. Identifying the presence of an anomaly alone does not provide a complete understanding of the incident.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The third objective is to determine the precise temporal interval in which anomalous behaviour occurs. This enables a more detailed understanding of how network incidents develop, how long they persist and when normal operating conditions are restored.
</p>

<div style="border-top: 1px solid #333333; margin: 24px 0;"></div>

<h3 style="color: #A50044; font-family: Arial, sans-serif; font-size: 17px; letter-spacing: 1px;">
04 · AFFECTED KPI IDENTIFICATION
</h3>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
A network anomaly can affect multiple performance indicators simultaneously. For example, radio interference may increase transmission errors and reduce signal quality, while network congestion can lead to higher resource utilization and degraded data transmission performance. Identifying which Key Performance Indicators (KPIs) are affected is therefore essential for understanding the operational impact of an incident.
</p>

<p style="color: #FFFFFF; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The fourth objective is to identify the KPI channels associated with each anomalous event. This multi-label classification task aims to determine which network performance indicators are affected, providing additional information for network diagnosis, fault isolation and root cause analysis.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-top: 0;">
<a href="https://huggingface.co/datasets/AliMaatouk/TelecomTS" target="_blank" style="color: #8AB9FF; text-decoration: none;">TelecomTS</a> is a multimodal dataset designed for the analysis of telecommunications networks through time-series measurements and natural language information. It provides a combination of network performance indicators, anomaly annotations, contextual metadata and textual descriptions, offering a comprehensive representation of different network operating conditions.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The dataset contains 32,000 samples organized into observation windows. Each sample includes 18 KPI channels, with 128 consecutive measurements per channel, collected at a sampling frequency of 10 Hz. In addition to numerical measurements, the dataset incorporates categorical and integer-valued information describing different aspects of the network environment.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
A relevant component of TelecomTS is its anomaly annotation system, which identifies whether an anomalous event is present, specifies its category and provides information about its temporal duration and potentially affected KPIs. The dataset covers 11 anomaly categories, including interference, jamming, congestion, hardware failures and resource allocation problems. Both normal and anomalous network conditions are represented.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-bottom: 0;">
Beyond structured measurements and annotations, TelecomTS also incorporates natural language descriptions and troubleshooting-related information. These elements provide additional context about network scenarios and anomalous behavior, making the dataset particularly interesting for multimodal network observability and root cause analysis.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-top: 0;">
TelecomTS contains several types of variables that can be organized according to their role in the supervised learning tasks developed in this project. The main distinction is between predictive variables, which provide the information used by the models to make predictions, and target variables, which contain the ground-truth information used during training and evaluation.
</p>

<h4 style="color: #A50044; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Predictive Variables (Features)</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The primary predictive variables are the 18 KPI channels, each containing 128 consecutive measurements per observation window. These time-series measurements describe network behavior and will be used as the main input for the predictive models. Depending on the task, they can be processed directly as multivariate time series or transformed into statistical features.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The dataset also includes contextual information, such as integer-valued and categorical variables describing network conditions or scenarios. These variables may provide additional predictive information, provided that they are available at inference time and do not directly reveal the target. Their inclusion will therefore depend on an individual assessment of their meaning and potential information leakage.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #004D98; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Target Variables (Labels)</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
The anomaly annotations provide the ground-truth information required to define the supervised learning targets. Different fields within these annotations can be used to construct distinct prediction tasks:
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">1. Binary Anomaly Detection:</strong> The <code style="color: #8AB9FF;">anomalies.exists</code> field acts as the target variable (y), indicating whether an observation window contains an anomaly (1) or represents normal network behavior (0). The objective is to identify anomalous network conditions.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">2. Anomaly Type Classification:</strong> The <code style="color: #8AB9FF;">anomalies.type</code> field acts as the target variable (y), identifying the category of an anomalous event. This task is performed on anomalous observations and aims to classify the underlying anomaly type into one of the 11 available categories.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-bottom: 0;">
<strong style="color: #FFFFFF;">3. Temporal Anomaly Localization:</strong> The <code style="color: #8AB9FF;">anomalies.anomaly_duration.start</code> and <code style="color: #8AB9FF;">anomalies.anomaly_duration.end</code> fields are used to construct a sequence-level target (y) of 128 binary values. Each position indicates whether an anomaly is present at that timestep. The objective is to identify not only whether an anomaly exists, but also when it occurs within the observation window.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">4. Affected KPI Identification:</strong> The <code style="color: #8AB9FF;">anomalies.affected_kpis</code> field is used as the target variable (y) to identify which KPI channels are affected by an anomalous event. Since a single anomaly can affect multiple KPIs simultaneously, this task is formulated as a multi-label classification problem. The objective is to identify the network performance indicators associated with each anomaly, providing additional information for fault isolation and root cause analysis.
</p>

<div style="background: linear-gradient(135deg, #050505 0%, #071426 55%, #18000C 100%); border: 1px solid #303030; border-radius: 16px; padding: 24px;">

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-top: 0;">
The <code style="color: #8AB9FF;">KPIs</code> column in our DataFrame contains 18 different Key Performance Indicators (KPIs) that describe various aspects of telecommunications network performance. These variables cover radio signal quality, resource allocation, traffic volume, transmission reliability and communication protocols. Understanding what each KPI represents from a telecommunications perspective is essential before proceeding with the exploratory data analysis and developing the supervised learning models. The following sections describe each variable and its role in monitoring network behavior.
</p>

<h4 style="color: #A50044; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-top: 0; margin-bottom: 10px;">Downlink (DL) KPIs</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">1. DL_BLER (Downlink Block Error Rate):</strong> Measures the proportion of downlink transport blocks that are received incorrectly by the User Equipment (UE). A high BLER indicates poor transmission quality and may trigger retransmissions through mechanisms such as HARQ, reducing spectral efficiency and potentially affecting user experience.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">2. DL_MCS (Downlink Modulation and Coding Scheme):</strong> Represents the modulation and coding configuration selected for downlink transmissions. Higher MCS values generally allow more information to be transmitted per resource block, but require better radio conditions to maintain reliable communication. Lower values are typically selected when channel quality deteriorates.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">3. DL_NumberOfPackets:</strong> Represents the number of packets transmitted through the downlink direction, from the base station towards the UE. It provides information about traffic volume and network activity, helping identify changes in data demand or unusual traffic patterns.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">4. DL_Protocol:</strong> Identifies the protocol associated with downlink traffic or transmission. This categorical variable provides information about the type of communication being handled by the network and may help distinguish different traffic patterns or application behaviors.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #004D98; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Uplink (UL) KPIs</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">5. Estimated_UL_Buffer:</strong> Represents an estimate of the amount of uplink data waiting to be transmitted by the UE. Buffer information is relevant to radio resource scheduling because the base station uses information about pending data to allocate uplink transmission resources. Increasing buffer levels may indicate growing traffic demand or insufficient uplink capacity.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">6. UL_BLER (Uplink Block Error Rate):</strong> Measures the proportion of uplink transport blocks that are received incorrectly by the base station. High values indicate unreliable uplink transmissions and can lead to retransmissions, reduced throughput and increased resource consumption.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">7. UL_MCS (Uplink Modulation and Coding Scheme):</strong> Represents the modulation and coding configuration used for uplink transmissions from the UE to the base station. Higher values generally provide greater transmission efficiency, while lower values are usually associated with more conservative configurations under poorer radio conditions.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">8. UL_NPRB (Uplink Number of Physical Resource Blocks):</strong> Indicates the number of Physical Resource Blocks allocated or used for uplink transmission. PRBs are the fundamental time-frequency resource units assigned to users in an OFDMA-based radio system. This KPI provides information about uplink resource allocation and network load.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">9. UL_NumberOfPackets:</strong> Represents the number of packets transmitted in the uplink direction, from the UE towards the base station. It helps characterize user-generated traffic and identify variations in uplink activity.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">10. UL_Protocol:</strong> Identifies the protocol associated with uplink traffic or transmission. It provides contextual information about the communication being handled and may help distinguish different types of network activity.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #A50044; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Radio Resource Utilization KPIs</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">11. PRB_Utilization_DL (Downlink PRB Utilization):</strong> Measures the proportion of available downlink Physical Resource Blocks being used by the network. High utilization indicates that a large share of the available radio resources is occupied, potentially leading to congestion, reduced throughput and limited capacity for additional users.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">12. PRB_Utilization_UL (Uplink PRB Utilization):</strong> Measures the proportion of available uplink Physical Resource Blocks being used. It reflects the demand for uplink radio resources and can help identify uplink congestion or inefficient resource allocation.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">13. PRBs_DL_Current:</strong> Represents the current number of Physical Resource Blocks assigned or used in the downlink direction. Unlike utilization, which expresses resource consumption relative to available capacity, this variable provides an absolute resource count.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #004D98; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Radio Signal Quality KPIs</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">14. RSRP (Reference Signal Received Power):</strong> Measures the average received power of specific reference signals transmitted by the base station. It is commonly expressed in dBm and is used to evaluate radio coverage and signal strength. More negative values generally indicate weaker received signals, which may affect connectivity and transmission performance.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">15. UL_SNR (Uplink Signal-to-Noise Ratio):</strong> Measures the ratio between the received uplink signal power and the noise power at the receiver. A higher SNR generally indicates better signal quality and supports more efficient modulation and coding configurations. Low values may result in transmission errors and reduced uplink performance.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #A50044; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Traffic Volume KPIs</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">16. RX_Bytes (Received Bytes):</strong> Represents the volume of data received by the monitored network entity during the observation interval. It provides information about incoming traffic and can be used to analyze traffic intensity, variations in data consumption and potential abnormal traffic patterns. The exact direction depends on the measurement point used in the dataset.
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">17. TX_Bytes (Transmitted Bytes):</strong> Represents the volume of data transmitted by the monitored network entity during the observation interval. It helps characterize outgoing traffic and understand how data demand evolves over time. As with RX_Bytes, its precise direction depends on the monitored interface or entity.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #004D98; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Additional Resource Allocation KPI</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-top: 0;">
The <code style="color: #8AB9FF;">anomalies</code> column in our DataFrame contains five fields describing the presence, characteristics and potential causes of anomalous events within the telecommunications network. These annotations provide the ground-truth information required to develop and evaluate the supervised learning models. They include binary labels, anomaly categories, temporal information, affected network indicators and troubleshooting information. Understanding each field is essential to define the different prediction objectives and avoid data leakage during model training.
</p>

<h4 style="color: #A50044; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Anomaly Annotation Variables</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">1. exists:</strong> Indicates whether an anomalous event is present in the observation window. It is a binary variable that distinguishes normal network behavior from anomalous conditions. <strong style="color: #E6A0B4;">Target variable (y) for Binary Anomaly Detection.</strong>
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">2. type:</strong> Identifies the category of the detected anomaly, such as jamming, interference, network congestion or hardware failures. This categorical variable provides information about the nature of the network problem. <strong style="color: #E6A0B4;">Target variable (y) for Anomaly Type Classification.</strong>
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">3. anomaly_duration:</strong> Contains the temporal boundaries of the anomalous event within the observation window, represented by its <code style="color: #8AB9FF;">start</code> and <code style="color: #8AB9FF;">end</code> positions. This information allows us to determine which time steps are affected by the anomaly. <strong style="color: #E6A0B4;">Used to construct the sequence-level target (y) for Temporal Anomaly Localization.</strong>
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8;">
<strong style="color: #FFFFFF;">4. affected_kpis:</strong> Identifies the Key Performance Indicators associated with or affected by the anomalous event. It provides information about which network performance metrics are involved in the anomaly. <strong style="color: #E6A0B4;">Potential target variable (y) for Multi-label Affected KPI Identification.</strong>
</p>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-bottom: 0;">
<strong style="color: #FFFFFF;">5. troubleshooting_tickets:</strong> Contains troubleshooting-related information associated with network anomalies. These records can provide additional context about the diagnosis, investigation or resolution of network problems.
</p>

<div style="border-top: 1px solid #333333; margin: 22px 0;"></div>

<h4 style="color: #004D98; font-family: Arial, sans-serif; font-size: 16px; letter-spacing: 0.5px; margin-bottom: 10px;">Role in the Supervised Learning Project</h4>

<p style="color: #D0D0D0; font-family: Arial, sans-serif; font-size: 14px; line-height: 1.8; margin-bottom: 0;">
The <code style="color: #8AB9FF;">anomalies</code> fields will primarily be used to construct the ground-truth labels for the four supervised learning objectives: binary anomaly detection, anomaly type classification, temporal anomaly localization and affected KPI identification. These annotation fields must remain separate from the predictive inputs whenever they define the target of a given task. In particular, using them as model inputs could introduce data leakage, allowing the model to access information that would not be available when making real-world predictions.
</p>

2.1 · BINARY ANOMALY DETECTION DATASET

<p>

        <li><strong style="color: #A50044;">1:</strong> Anomalous network behavior.</li>

<p>
        The objective is to train a supervised model capable of distinguishing normal network conditions from anomalous ones using the temporal patterns of the KPIs.

2.2 · ANOMALY TYPE CLASSIFICATION DATASET

<div style="background: linear-gradient(135deg, #050505 0%, #071426 55%, #18000C 100%); border-left: 4px solid #A50044; padding: 18px 22px; border-radius: 8px; font-family: Arial, sans-serif; color: #D0D0D0; font-size: 14px; line-height: 1.8; margin-bottom: 18px;">


<p>
        This dataset focuses exclusively on anomalous network observations. It contains <strong style="color: #FFFFFF;">1,235 samples</strong> with a valid anomaly type annotation.
</p>

<p style="color: #E6A0B4; font-weight: bold;">Input features (X)</p>

<p>
        The input tensor has a shape of <code style="color: #E6A0B4;">(1235, 128, 24)</code>. Each sample contains 128 timesteps and 24 input features describing the network behavior.
</p>

<p style="color: #E6A0B4; font-weight: bold;">Target (y)</p>

<p>
        The target is a one-dimensional array containing 1,235 categorical labels. Each label identifies the type of anomaly affecting the corresponding network observation. The dataset includes 11 different anomaly classes.

<p>
        Unlike binary anomaly detection, this task does not distinguish between normal and anomalous behavior. Instead, it determines <strong style="color: #FFFFFF;">which type of anomaly</strong> is present in an already identified anomalous observation.

<p>
        The objective is to train a multiclass classification model capable of recognizing different anomaly patterns from their KPI time series.
</p>

</div>

```python
X_type = datasets["anomaly_type"]["X"]
y_type = datasets["anomaly_type"]["y"]
type_positions = datasets["anomaly_type"]["positions"]

(
    X_type_train,
    X_type_test,
    y_type_train,
    y_type_test,
    type_positions_train,
    type_positions_test,
) = train_test_split(
    X_type,
    y_type,
    type_positions,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y_type,
)
```

2.3 · TEMPORAL ANOMALY LOCALIZATION DATASET

<div style="background: linear-gradient(135deg, #050505 0%, #071426 55%, #18000C 100%); border-left: 4px solid #004D98; padding: 18px 22px; border-radius: 8px; font-family: Arial, sans-serif; color: #D0D0D0; font-size: 14px; line-height: 1.8; margin-bottom: 18px;">


<p>
        This dataset is designed to identify not only whether an anomaly occurs, but also <strong style="color: #FFFFFF;">when it occurs within the observation window</strong>. It contains the same 1,235 anomalous samples used for anomaly type classification.
</p>

<p style="color: #6CABDD; font-weight: bold;">Input features (X)</p>

<p>
        The input tensor has a shape of <code style="color: #6CABDD;">(1235, 128, 24)</code>. Each observation contains 128 timesteps and 24 KPI-related input features.
</p>

<p style="color: #6CABDD; font-weight: bold;">Target (y)</p>

<p>
        The target has a shape of <code style="color: #6CABDD;">(1235, 128)</code>. Unlike the previous tasks, each sample is associated with a sequence of 128 binary labels, one for each timestep:
</p>

<ul>
        <li><strong style="color: #6CABDD;">0:</strong> Timestep outside the annotated anomaly interval.</li>
        <li><strong style="color: #E6A0B4;">1:</strong> Timestep inside the annotated anomaly interval.</li>
</ul>

<p>
        These labels are constructed from the <code style="color: #6CABDD;">anomaly_duration.start</code> and <code style="color: #6CABDD;">anomaly_duration.end</code> annotations.
</p>

<p>
        The objective is to train a sequence-labeling model capable of identifying the temporal boundaries of anomalous behavior within each observation window.
</p>

</div>

```python
X_temporal = datasets["temporal_localization"]["X"]
y_temporal = datasets["temporal_localization"]["y"]

temporal_positions = datasets["temporal_localization"]["positions"]

# Preserve metadata alignment when creating the task-specific split.
temporal_types = np.array([
    df.iloc[position]["anomalies"]["type"]
    for position in temporal_positions
])

(
    X_temporal_train,
    X_temporal_test,
    y_temporal_train,
    y_temporal_test,
    temporal_types_train,
    temporal_types_test,
    temporal_positions_train,
    temporal_positions_test,
) = train_test_split(
    X_temporal,
    y_temporal,
    temporal_types,
    temporal_positions,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=temporal_types,
)
```

2.4 · AFFECTED KPI IDENTIFICATION DATASET

<div style="height: 2px; width: 90px; margin-top: 14px; background: linear-gradient(90deg, #004D98, #A50044); border-radius: 5px;"></div>

</div>

<div style="background: linear-gradient(135deg, #050505 0%, #071426 55%, #18000C 100%); border-left: 4px solid #A50044; padding: 18px 22px; border-radius: 8px; font-family: Arial, sans-serif; color: #D0D0D0; font-size: 14px; line-height: 1.8; margin-bottom: 18px;">

<p>
        This dataset aims to identify which Key Performance Indicators (KPIs) are affected by an anomaly. It contains <strong style="color: #FFFFFF;">1,235 anomalous samples</strong> with valid affected KPI annotations.
</p>

<p style="color: #E6A0B4; font-weight: bold;">Input features (X)</p>

<p>
        The input tensor has a shape of <code style="color: #E6A0B4;">(1235, 128, 24)</code>. Each sample contains 128 timesteps and 24 input features describing network behavior.
</p>

<p style="color: #E6A0B4; font-weight: bold;">Target (y)</p>

<p>
        The target matrix has a shape of <code style="color: #E6A0B4;">(1235, 16)</code>. Each row represents one anomalous observation, while each of the 16 columns corresponds to an affected KPI class.
</p>

<p>
        The labels are encoded using <code style="color: #E6A0B4;">MultiLabelBinarizer</code>, producing a multi-label binary representation:
</p>

<ul>
        <li><strong style="color: #6CABDD;">1:</strong> The corresponding KPI is annotated as affected.</li>
        <li><strong style="color: #E6A0B4;">0:</strong> The corresponding KPI is not included in the affected KPI annotations.</li>
</ul>

<p>
        Since an anomaly can affect several KPIs simultaneously, this is a <strong style="color: #FFFFFF;">multi-label classification problem</strong>, rather than a conventional multiclass classification problem.
</p>

<p>
        The objective is to identify the KPIs associated with anomalous behavior, providing useful information for fault isolation and subsequent root cause analysis.
</p>

</div>

```python
X_affected = datasets["affected_kpis"]["X"]
y_affected = datasets["affected_kpis"]["y"]
affected_positions = datasets["affected_kpis"]["positions"]
affected_types = np.array([
    df.iloc[position]["anomalies"]["type"]
    for position in affected_positions
])

multilabel_splitter = MultilabelStratifiedShuffleSplit(
    n_splits=1,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

train_indices, test_indices = next(
    multilabel_splitter.split(X_affected, y_affected)
)

X_affected_train = X_affected[train_indices]
X_affected_test = X_affected[test_indices]

y_affected_train = y_affected[train_indices]
y_affected_test = y_affected[test_indices]
affected_types_train = affected_types[train_indices]
affected_types_test = affected_types[test_indices]
affected_positions_train = affected_positions[train_indices]
affected_positions_test = affected_positions[test_indices]
```
